```python
import numpy as np

# OR Gate Dataset
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([0,1,1,1])

learning_rate = 0.1
epochs = 10

weights = np.zeros(2)
bias = 0

def step(x):
    if x >= 0:
        return 1
    return 0

print("Initial Weights:", weights)
print("Initial Bias:", bias)

update = 1

for epoch in range(epochs):

    print(f"\n========== Epoch {epoch+1} ==========")

    errors = 0

    for xi, target in zip(X, y):

        z = np.dot(weights, xi) + bias

        prediction = step(z)

        error = target - prediction

        if error != 0:

            weights = weights + learning_rate * error * xi
            bias = bias + learning_rate * error

            errors += 1

            print("\nUpdate", update)
            print("Input :", xi)
            print("Target :", target)
            print("Prediction :", prediction)
            print("Weights :", weights)
            print("Bias :", bias)

            update += 1

    print("Errors =", errors)

    if errors == 0:
        print("\nTraining Converged")
        break

print("\nFinal Weights :", weights)
print("Final Bias :", bias)

print("\nFinal Predictions")

for sample in X:

    pred = step(np.dot(weights, sample) + bias)

    print(sample, " ---> ", pred)
```

```python
import numpy as np
import matplotlib.pyplot as plt

X_xor = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
], dtype=float)

y_xor = np.array([
    0,
    1,
    1,
    0
], dtype=float)

class MLP:

    def __init__(self,
                 n_input=2,
                 n_hidden=2,
                 n_output=1,
                 learning_rate=0.5,
                 seed=42):

        rng = np.random.default_rng(seed)

        self.W1 = rng.uniform(-1,1,(n_input,n_hidden))
        self.b1 = np.zeros(n_hidden)

        self.W2 = rng.uniform(-1,1,(n_hidden,n_output))
        self.b2 = np.zeros(n_output)

        self.lr = learning_rate

        self.weight_history = []


    def sigmoid(self,x):

        return 1/(1+np.exp(-x))


    def sigmoid_derivative(self,x):

        return x*(1-x)


    def forward(self,X):

        self.z1 = np.dot(X,self.W1)+self.b1

        self.a1 = self.sigmoid(self.z1)

        self.z2 = np.dot(self.a1,self.W2)+self.b2

        self.output = self.sigmoid(self.z2)

        return self.output


    def predict(self,X):

        output = self.forward(X)

        return (output>=0.5).astype(int).flatten()

    def train(self,
              X,
              y,
              epochs=5000,
              log_every=1):

        y = y.reshape(-1,1)

        for epoch in range(1,epochs+1):

            output = self.forward(X)

            error = y-output

            loss = np.mean(error**2)

            d_output = error*self.sigmoid_derivative(output)

            error_hidden = np.dot(d_output,self.W2.T)

            d_hidden = error_hidden*self.sigmoid_derivative(self.a1)

            self.W2 += self.lr*np.dot(self.a1.T,d_output)

            self.b2 += self.lr*np.sum(d_output,axis=0)

            self.W1 += self.lr*np.dot(X.T,d_hidden)

            self.b1 += self.lr*np.sum(d_hidden,axis=0)

            if epoch==1 or epoch%log_every==0:

                self.weight_history.append(

                    (
                        epoch,

                        self.W1.copy(),

                        self.b1.copy(),

                        self.W2.copy(),

                        self.b2.copy(),

                        loss
                    )

                )

                print("\n"+"="*60)

                print("Epoch :",epoch)

                print("Loss :",loss)

                print("\nW1")

                print(np.round(self.W1,3))

                print("\nb1")

                print(np.round(self.b1,3))

                print("\nW2")

                print(np.round(self.W2,3))

                print("\nb2")

                print(np.round(self.b2,3))

            if loss<0.0001:

                print("\nTraining Converged")

                break
```

```python
def plot_mlp_decision_boundary(model, W1, b1, W2, b2,
                               X, y,
                               epoch,
                               loss):

    # Backup current weights
    old_W1 = model.W1.copy()
    old_b1 = model.b1.copy()
    old_W2 = model.W2.copy()
    old_b2 = model.b2.copy()

    # Load stored weights
    model.W1 = W1.copy()
    model.b1 = b1.copy()
    model.W2 = W2.copy()
    model.b2 = b2.copy()

    # Create mesh grid
    x_min, x_max = -0.5, 1.5
    y_min, y_max = -0.5, 1.5

    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 250),
        np.linspace(y_min, y_max, 250)
    )

    grid = np.c_[xx.ravel(), yy.ravel()]

    # Predict every point
    Z = model.forward(grid)

    Z = Z.reshape(xx.shape)

    plt.figure(figsize=(6,6))

    # Background probability
    plt.contourf(
        xx,
        yy,
        Z,
        levels=40,
        cmap='coolwarm',
        alpha=0.7
    )

    # Decision Boundary
    plt.contour(
        xx,
        yy,
        Z,
        levels=[0.5],
        colors='black',
        linewidths=2
    )

    # Training Samples
    plt.scatter(
        X[y==0][:,0],
        X[y==0][:,1],
        color='red',
        s=150,
        edgecolors='black',
        label='Class 0'
    )

    plt.scatter(
        X[y==1][:,0],
        X[y==1][:,1],
        color='blue',
        s=150,
        edgecolors='black',
        label='Class 1'
    )

    plt.title(f"Decision Boundary\nEpoch {epoch}   Loss = {loss:.5f}")

    plt.xlabel("Input x1")
    plt.ylabel("Input x2")

    plt.xlim(-0.5,1.5)
    plt.ylim(-0.5,1.5)

    plt.grid(True)

    plt.legend()

    plt.show()

    # Restore original weights
    model.W1 = old_W1
    model.b1 = old_b1
    model.W2 = old_W2
    model.b2 = old_b2
```

```python
print("="*60)
print("XOR GATE (Multi-Layer Perceptron)")
print("="*60)

mlp = MLP(
    n_input=2,
    n_hidden=2,
    n_output=1,
    learning_rate=0.5
)

mlp.train(
    X_xor,
    y_xor,
    epochs=5000,
    log_every=500
)

print("\nFinal Predictions")

print(mlp.predict(X_xor))

print("\nExpected")

print(y_xor.astype(int))

for epoch, W1, b1, W2, b2, loss in mlp.weight_history:

    plot_mlp_decision_boundary(
        mlp,
        W1,
        b1,
        W2,
        b2,
        X_xor,
        y_xor,
        epoch,
        loss
    )
```
